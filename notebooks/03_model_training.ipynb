```python
# ============================================================
# IMPORTS
# ============================================================

from pathlib import Path
import torch

from ultralytics import YOLO

print("Torch Version:", torch.__version__)
print("CUDA Available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
# ============================================================
# PATHS
# ============================================================

PROJECT_ROOT = Path("../")

DATASET_PATH = (
    PROJECT_ROOT /
    "datasets" /
    "processed" /
    "dataset_final"
)

YAML_PATH = (
    DATASET_PATH /
    "data.yaml"
)

print(YAML_PATH)
```

```python
# ============================================================
# VALIDATE MODEL
# ============================================================

metrics = model.val()

print(metrics)
```

```python
# ============================================================
# LOAD BEST TRAINED MODEL
# ============================================================

BEST_MODEL_PATH = (
    PROJECT_ROOT /
    "models" /
    "trained" /
    "ppe_yolo11m_best.pt"
)

print("Best Model Path:")
print(BEST_MODEL_PATH)

best_model = YOLO(str(BEST_MODEL_PATH))

print("\nModel loaded successfully.")
```

```python
# ============================================================
# FULL MODEL EVALUATION
# ============================================================

metrics = best_model.val(

    data=str(YAML_PATH),

    split="test",

    imgsz=IMAGE_SIZE,

    conf=0.25,

    iou=0.6,

    plots=True
)

print("\nValidation completed.")
```

```python
# ============================================================
# PERFORMANCE METRICS
# ============================================================

precision = metrics.box.mp
recall = metrics.box.mr
map50 = metrics.box.map50
map5095 = metrics.box.map

# Approximate Detection Accuracy
accuracy = (precision + recall) / 2

print("\n" + "="*70)
print("FINAL MODEL PERFORMANCE")
print("="*70)

print(f"\nApprox Accuracy : {accuracy*100:.2f}%")

print(f"\nPrecision       : {precision*100:.2f}%")

print(f"Recall          : {recall*100:.2f}%")

print(f"mAP50           : {map50*100:.2f}%")

print(f"mAP50-95        : {map5095*100:.2f}%")
```

```python
# ============================================================
# CLASS-WISE PERFORMANCE
# ============================================================

CLASS_NAMES = [

    "Hardhat",
    "Mask",
    "Person",
    "Safety Cone",
    "Safety Vest",
    "machinery",
    "vehicle"
]

print("\n" + "="*70)
print("CLASS-WISE PERFORMANCE")
print("="*70)

for i, class_name in enumerate(CLASS_NAMES):

    try:

        ap50 = metrics.box.ap50[i]

        ap = metrics.box.ap[i]

        print(f"\n{class_name}")

        print(f"mAP50     : {ap50*100:.2f}%")

        print(f"mAP50-95  : {ap*100:.2f}%")

    except:
        print(f"\n{class_name} -> Metric unavailable")
```

```python
# ============================================================
# SHOW TRAINING RESULTS
# ============================================================

from PIL import Image
import matplotlib.pyplot as plt

results_dir = ROOT / PROJECT_NAME / RUN_NAME

images_to_show = [

    "results.png",
    "PR_curve.png",
    "F1_curve.png",
    "confusion_matrix.png"
]

fig, axes = plt.subplots(2, 2, figsize=(18, 14))

for ax, image_name in zip(axes.flatten(), images_to_show):

    image_path = results_dir / image_name

    if image_path.exists():

        img = Image.open(image_path)

        ax.imshow(img)

        ax.set_title(image_name)

    else:

        ax.text(
            0.5,
            0.5,
            f"{image_name}\nNot Found",
            ha="center",
            va="center"
        )

    ax.axis("off")

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# TEST MODEL INFERENCE
# ============================================================

import random
import cv2

test_images = random.sample(
    list((DATASET_PATH / "test" / "images").glob("*.jpg")),
    4
)

fig, axes = plt.subplots(2, 2, figsize=(18, 18))

for ax, image_path in zip(axes.flatten(), test_images):

    results = best_model.predict(

        source=str(image_path),

        conf=0.35,

        imgsz=IMAGE_SIZE
    )

    plotted = results[0].plot()

    plotted = cv2.cvtColor(
        plotted,
        cv2.COLOR_BGR2RGB
    )

    ax.imshow(plotted)

    ax.set_title(image_path.name)

    ax.axis("off")

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# FPS BENCHMARK
# ============================================================

import time
import random

sample_image = random.choice(
    list((DATASET_PATH / "test" / "images").glob("*.jpg"))
)

# Warmup
for _ in range(5):

    _ = best_model.predict(
        source=str(sample_image),
        verbose=False
    )

# Benchmark
start = time.time()

num_runs = 50

for _ in range(num_runs):

    _ = best_model.predict(
        source=str(sample_image),
        verbose=False
    )

end = time.time()

total_time = end - start

fps = num_runs / total_time

print("\n" + "="*60)
print("INFERENCE PERFORMANCE")
print("="*60)

print(f"\nFPS : {fps:.2f}")
```
